```python
model_features = [
    'amount_log',
    'transaction type',
    'merchant_category',
    'sender_age_group',
    'receiver_age_group',
    'sender_state',
    'sender_bank',
    'receiver_bank',
    'device_type',
    'network_type',
    'hour_of_day',
    'day_of_week',
    'is_weekend',
    'time_period',
    'weekend_evening',
    'high_value_flag'
]

X = df[model_features]
y = df['fraud_flag']

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_features = [
    'transaction type',
    'merchant_category',
    'sender_age_group',
    'receiver_age_group',
    'sender_state',
    'sender_bank',
    'receiver_bank',
    'device_type',
    'network_type',
    'day_of_week',
    'time_period'
]

numerical_features = [
    'amount_log',
    'hour_of_day',
    'is_weekend',
    'weekend_evening',
    'high_value_flag'
]

preprocessor = ColumnTransformer(
    transformers=[
        ('categorical', OneHotEncoder(handle_unknown='ignore'), categorical_features),
        ('numerical', 'passthrough', numerical_features)
    ]
)

print("Categorical features:", len(categorical_features))
print("Numerical features:", len(numerical_features))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('classifier', LogisticRegression(
            class_weight='balanced',
            max_iter=1000,
            random_state=42
        ))
    ]
)

print("Logistic Regression pipeline created successfully.")
```

```python
logistic_model.fit(X_train, y_train)

print("Logistic Regression model trained successfully.")
```

```python
y_pred = logistic_model.predict(X_test)

y_proba = logistic_model.predict_proba(X_test)[:, 1]

print("Predictions generated successfully.")
print("Predicted fraud cases:", y_pred.sum())
```

```python
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    roc_auc_score,
    average_precision_score
)

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

print("\nClassification Report:")
print(classification_report(y_test, y_pred, digits=4))

print("ROC-AUC:", round(roc_auc_score(y_test, y_proba), 4))
print("PR-AUC:", round(average_precision_score(y_test, y_proba), 4))
```

```python
from sklearn.ensemble import RandomForestClassifier

random_forest_model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('classifier', RandomForestClassifier(
            n_estimators=200,
            class_weight='balanced',
            random_state=42,
            n_jobs=-1
        ))
    ]
)

print("Random Forest pipeline created successfully.")
```

```python
random_forest_model.fit(X_train, y_train)

print("Random Forest model trained successfully.")
```

```python
rf_pred = random_forest_model.predict(X_test)

rf_proba = random_forest_model.predict_proba(X_test)[:, 1]

print("Predictions generated successfully.")
print("Predicted fraud cases:", rf_pred.sum())
```

```python
rf_cm = confusion_matrix(y_test, rf_pred)

print("Confusion Matrix:")
print(rf_cm)

print("\nClassification Report:")
print(classification_report(y_test, rf_pred, digits=4))

print("ROC-AUC:", round(roc_auc_score(y_test, rf_proba), 4))
print("PR-AUC:", round(average_precision_score(y_test, rf_proba), 4))
```

```python
print("Minimum probability:", rf_proba.min())
print("Maximum probability:", rf_proba.max())
print("Mean probability:", rf_proba.mean())

print("\nHighest 20 fraud probabilities:")
print(sorted(rf_proba, reverse=True)[:20])
```

```python
import numpy as np

normal_proba = rf_proba[y_test == 0]
fraud_proba = rf_proba[y_test == 1]

print("Normal transactions:")
print("Mean probability:", normal_proba.mean())
print("Median probability:", np.median(normal_proba))

print("\nActual fraud transactions:")
print("Mean probability:", fraud_proba.mean())
print("Median probability:", np.median(fraud_proba))

print("\nHighest probability among actual fraud:")
print(fraud_proba.max())

print("\nHighest probability among normal:")
print(normal_proba.max())
```

```python
overall_rate = df['fraud_flag'].mean()

risk_indicators = {
    'High Value (>₹5,000)': df['high_value_flag'] == 1,
    'Recharge': df['transaction type'] == 'Recharge',
    'WiFi': df['network_type'] == 'WiFi',
    'Weekend': df['is_weekend'] == 1,
    'Evening': df['time_period'] == 'Evening'
}

for name, condition in risk_indicators.items():
    rate = df.loc[condition, 'fraud_flag'].mean()
    relative_risk = rate / overall_rate

    print(
        f"{name}: "
        f"Fraud rate = {rate:.4%}, "
        f"Relative risk = {relative_risk:.2f}x"
    )
```

```python
df['risk_score'] = (
    df['high_value_flag'] * 3
    + (df['transaction type'] == 'Recharge').astype(int) * 2
    + (df['network_type'] == 'WiFi').astype(int) * 2
    + df['is_weekend'] * 1
    + (df['time_period'] == 'Evening').astype(int) * 1
)

print(df['risk_score'].value_counts().sort_index())
```

```python
risk_analysis = (
    df.groupby('risk_score')['fraud_flag']
      .agg(
          total_transactions='count',
          fraud_cases='sum',
          fraud_rate='mean'
      )
      .reset_index()
)

risk_analysis['fraud_rate_pct'] = risk_analysis['fraud_rate'] * 100

print(risk_analysis)
```

```python
print("\nRisk score summary:")
print(
    risk_analysis[
        ['risk_score', 'total_transactions', 'fraud_cases', 'fraud_rate_pct']
    ].to_string(index=False)
)
```

```python
high_risk = df[df['risk_score'] >= 3]

print("High-risk transactions:", len(high_risk))
print("High-risk fraud cases:", high_risk['fraud_flag'].sum())

print(
    "High-risk fraud rate:",
    round(high_risk['fraud_flag'].mean() * 100, 4),
    "%"
)

print(
    "Share of all transactions:",
    round(len(high_risk) / len(df) * 100, 2),
    "%"
)

print(
    "Share of all fraud cases captured:",
    round(
        high_risk['fraud_flag'].sum() / df['fraud_flag'].sum() * 100,
        2
    ),
    "%"
)
```

```python
df['risk_tier'] = pd.cut(
    df['risk_score'],
    bins=[-1, 1, 3, float('inf')],
    labels=['Low', 'Medium', 'High']
)

print(df['risk_tier'].value_counts().sort_index())
```

```python
tier_analysis = (
    df.groupby('risk_tier', observed=True)['fraud_flag']
      .agg(
          total_transactions='count',
          fraud_cases='sum',
          fraud_rate='mean'
      )
      .reset_index()
)

tier_analysis['fraud_rate_pct'] = tier_analysis['fraud_rate'] * 100

print(tier_analysis.to_string(index=False))
```
